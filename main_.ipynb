```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.model_selection import TimeSeriesSplit, train_test_split, cross_val_score, KFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import joblib
import warnings
warnings.filterwarnings('ignore')

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_rows', 100)
```

## LOAD AND MERGE DATA

```python
# Load datasets
yield_df = pd.read_csv('data/crop_yield.csv')
soil_df = pd.read_csv('data/state_soil_data.csv')
weather_df = pd.read_csv('data/state_weather_data_1997_2020.csv')

print(f"Yield data: {yield_df.shape}")
print(f"Soil data: {soil_df.shape}")
print(f"Weather data: {weather_df.shape}")

# Merge datasets
master_df = pd.merge(yield_df, weather_df, on=["state", "year"])
master_df = pd.merge(master_df, soil_df, on='state')

print(f"\n Master dataset created: {master_df.shape}")
print(f"\nColumns: {list(master_df.columns)}")

# first few rows
master_df.head()
```

## BASIC DATA EXPLORATION

```python
# Check for missing values
missing = master_df.isnull().sum()
if missing.sum() == 0:
    print("No missing values found!")
else:
    print(missing[missing > 0])

# Basic statistics
print("\n📈 Dataset Overview:")
print(f"Total records: {len(master_df):,}")
print(f"Date range: {master_df['year'].min()} - {master_df['year'].max()}")
print(f"Number of states: {master_df['state'].nunique()}")
print(f"Number of crops: {master_df['crop'].nunique()}")
print(f"Number of seasons: {master_df['season'].nunique()}")

# Show data types
print("\n📋 Data Types:")
print(master_df.dtypes)
```

## YIELD STATISTICS BY CROP

```python
# Calculate yield statistics by crop
crop_stats = master_df.groupby('crop')['yield'].agg([
    ('count', 'count'),
    ('mean', 'mean'),
    ('median', 'median'),
    ('std', 'std'),
    ('min', 'min'),
    ('max', 'max')
]).round(2)

crop_stats = crop_stats.sort_values('mean', ascending=False)

print("\n📊 Top 10 Highest Average Yield Crops:")
print(crop_stats.head(10))

print("\n📊 Bottom 10 Lowest Average Yield Crops:")
print(crop_stats.tail(10))
```

## OUTLIER DETECTION (IQR METHOD)

```python
# Calculate IQR bounds
Q1 = master_df['yield'].quantile(0.25)
Q3 = master_df['yield'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Detect outliers
yield_outliers = master_df[(master_df['yield'] < lower_bound) | 
                           (master_df['yield'] > upper_bound)]

print(f"\n📊 Yield Outlier Analysis:")
print(f"  Q1 (25th percentile): {Q1:.2f}")
print(f"  Median (50th percentile): {master_df['yield'].median():.2f}")
print(f"  Q3 (75th percentile): {Q3:.2f}")
print(f"  IQR: {IQR:.2f}")
print(f"  Lower Bound: {lower_bound:.2f}")
print(f"  Upper Bound: {upper_bound:.2f}")
print(f"\n  🚨 Total outliers detected: {len(yield_outliers)} ({len(yield_outliers)/len(master_df)*100:.2f}%)")

# Show outlier distribution by crop
print("\n📊 Outliers by Crop (Top 10):")
outlier_by_crop = yield_outliers.groupby('crop').size().sort_values(ascending=False)
print(outlier_by_crop.head(10))
```

## THREE OUTLIER HANDLING OPTIONS

```python
print("="*70)
print("🎯 OUTLIER HANDLING OPTIONS")
print("="*70)

# Calculate 99th percentile
percentile_99 = master_df['yield'].quantile(0.99)

print("\n📊 Three approaches to handle outliers:\n")

# Option 1: Remove outliers completely
df_option1 = master_df[(master_df['yield'] >= lower_bound) & 
                       (master_df['yield'] <= upper_bound)].copy()
print(f"✅ Option 1 - REMOVE ALL OUTLIERS (IQR Method):")
print(f"   Records removed: {len(master_df) - len(df_option1)} ({(len(master_df) - len(df_option1))/len(master_df)*100:.2f}%)")
print(f"   Remaining records: {len(df_option1):,}")
print(f"   New yield range: {df_option1['yield'].min():.2f} to {df_option1['yield'].max():.2f}")

# Option 2: Cap outliers at 99th percentile
df_option2 = master_df.copy()
df_option2['yield'] = df_option2['yield'].clip(upper=percentile_99)
capped_count = (master_df['yield'] > percentile_99).sum()
print(f"\n✅ Option 2 - CAP AT 99th PERCENTILE ({percentile_99:.2f}):")
print(f"   Records capped: {capped_count} ({capped_count/len(master_df)*100:.2f}%)")
print(f"   All records kept: {len(df_option2):,}")
print(f"   New yield range: {df_option2['yield'].min():.2f} to {df_option2['yield'].max():.2f}")

# Option 3: Keep all data
print(f"\n✅ Option 3 - KEEP ALL DATA (No Changes):")
print(f"   All records retained: {len(master_df):,}")
print(f"   Yield range: {master_df['yield'].min():.2f} to {master_df['yield'].max():.2f}")

print("\n💡 RECOMMENDATION: Option 2 (Cap at 99th percentile)")
print("   Reason: Balances data retention with outlier control")
```

## NORMALIZE FERTILIZER & PESTICIDE BY AREA

```python
print("\n" + "="*70)
print("📐 STEP 1: Normalize Fertilizer & Pesticide by Area")
print("="*70)

# Currently fertilizer/pesticide are TOTAL amounts
# We need them PER HECTARE for better predictions

print("\nBefore normalization:")
print(f"  Average fertilizer: {df['fertilizer'].mean():,.0f}")
print(f"  Average pesticide: {df['pesticide'].mean():,.0f}")
print(f"  Average area: {df['area'].mean():,.0f} hectares")

# Create normalized features
df['fertilizer_per_ha'] = df['fertilizer'] / (df['area'] + 1)  # +1 to avoid division by zero
df['pesticide_per_ha'] = df['pesticide'] / (df['area'] + 1)

print("\nAfter normalization (per hectare):")
print(f"  Average fertilizer per ha: {df['fertilizer_per_ha'].mean():,.2f}")
print(f"  Average pesticide per ha: {df['pesticide_per_ha'].mean():,.2f}")

print("\n✓ Created 2 new features: fertilizer_per_ha, pesticide_per_ha")
```

## CREATE NPK NUTRIENT FEATURES

```python
print("\n" + "="*70)
print("🌱 STEP 2: Create NPK Nutrient Features")
print("="*70)

# NPK ratios are important in agriculture
# Different crops need different nutrient balances

# Total NPK balance
df['NPK_total'] = df['N'] + df['P'] + df['K']

# NPK ratios (add small value to avoid division by zero)
df['N_to_P_ratio'] = df['N'] / (df['P'] + 0.1)
df['N_to_K_ratio'] = df['N'] / (df['K'] + 0.1)
df['P_to_K_ratio'] = df['P'] / (df['K'] + 0.1)

# NPK balance indicator (ideal ratio is roughly N:P:K = 3:1:2)
df['NPK_balance_score'] = np.abs((df['N']/3) - df['P']) + np.abs((df['N']/1.5) - df['K'])

print("\n✓ Created 5 NPK features:")
print("  • NPK_total")
print("  • N_to_P_ratio")
print("  • N_to_K_ratio") 
print("  • P_to_K_ratio")
print("  • NPK_balance_score")

print("\nNPK Feature Statistics:")
npk_features = ['NPK_total', 'N_to_P_ratio', 'N_to_K_ratio', 'P_to_K_ratio']
print(df[npk_features].describe().round(2))
```

## CREATE CLIMATE INTERACTION FEATURES

```python
print("\n" + "="*70)
print("🌤️ STEP 3: Create Climate Interaction Features")
print("="*70)

# Temperature and rainfall together affect crop growth
df['temp_rainfall_interaction'] = df['avg_temp_c'] * df['total_rainfall_mm']

# Temperature and humidity interaction
df['temp_humidity_interaction'] = df['avg_temp_c'] * df['avg_humidity_percent']

# Moisture availability index
df['moisture_index'] = (df['total_rainfall_mm'] * df['avg_humidity_percent']) / 100

# Growing degree days (simplified - base temp 10°C)
base_temp = 10
df['growing_degree_days'] = (df['avg_temp_c'] - base_temp).clip(lower=0)

# Rainfall category
df['rainfall_category'] = pd.cut(df['total_rainfall_mm'], 
                                  bins=[0, 500, 1000, 1500, 5000],
                                  labels=['Low', 'Medium', 'High', 'Very High'])

print("\n✓ Created 5 climate features:")
print("  • temp_rainfall_interaction")
print("  • temp_humidity_interaction")
print("  • moisture_index")
print("  • growing_degree_days")
print("  • rainfall_category")

print("\nClimate Feature Statistics:")
climate_features = ['temp_rainfall_interaction', 'moisture_index', 'growing_degree_days']
print(df[climate_features].describe().round(2))
```

## CREATE SOIL QUALITY FEATURES

```python
print("\n" + "="*70)
print("🌍 STEP 4: Create Soil Quality Features")
print("="*70)

# pH optimal range indicator (most crops prefer 6.0-7.5)
df['pH_optimal'] = ((df['pH'] >= 6.0) & (df['pH'] <= 7.5)).astype(int)

# pH category
df['pH_category'] = pd.cut(df['pH'], 
                            bins=[0, 5.5, 6.5, 7.5, 14],
                            labels=['Acidic', 'Slightly_Acidic', 'Neutral', 'Alkaline'])

# Soil fertility score (combination of NPK and pH)
df['soil_fertility_score'] = (
    (df['NPK_total'] / df['NPK_total'].max()) * 0.7 +  # 70% weight to NPK
    (df['pH_optimal']) * 0.3  # 30% weight to optimal pH
)

print("\n✓ Created 3 soil features:")
print("  • pH_optimal (1 if pH is between 6.0-7.5, else 0)")
print("  • pH_category")
print("  • soil_fertility_score")

print(f"\nSoil pH Distribution:")
print(df['pH_category'].value_counts())
```

### CREATE INPUT EFFICIENCY FEATURES

```python
print("\n" + "="*70)
print("⚡ STEP 5: Create Input Efficiency Features")
print("="*70)

# Fertilizer efficiency relative to rainfall
df['fertilizer_rainfall_ratio'] = df['fertilizer_per_ha'] / (df['total_rainfall_mm'] + 1)

# Pesticide efficiency per area
df['pesticide_efficiency'] = df['pesticide_per_ha'] / (df['area'] + 1)

# Input intensity (combined fertilizer and pesticide usage)
df['input_intensity'] = (
    (df['fertilizer_per_ha'] / df['fertilizer_per_ha'].max()) + 
    (df['pesticide_per_ha'] / df['pesticide_per_ha'].max())
) / 2

print("\n✓ Created 3 input efficiency features:")
print("  • fertilizer_rainfall_ratio")
print("  • pesticide_efficiency")
print("  • input_intensity")
```

## CREATE TEMPORAL FEATURES

```python
print("\n" + "="*70)
print("📅 STEP 6: Create Temporal Features")
print("="*70)

# Sort by state, crop, and year for lag features
df_sorted = df.sort_values(['state', 'crop', 'year']).reset_index(drop=True)

# Previous year's yield (lag feature)
df_sorted['yield_lag1'] = df_sorted.groupby(['state', 'crop'])['yield'].shift(1)

# 2-year moving average yield
df_sorted['yield_ma2'] = df_sorted.groupby(['state', 'crop'])['yield'].rolling(
    window=2, min_periods=1).mean().reset_index(drop=True)

# Year-over-year yield change
df_sorted['yield_change'] = df_sorted.groupby(['state', 'crop'])['yield'].diff()

# Years since start of data
df_sorted['years_since_start'] = df_sorted['year'] - df_sorted['year'].min()

print("\n✓ Created 4 temporal features:")
print("  • yield_lag1 (previous year's yield)")
print("  • yield_ma2 (2-year moving average)")
print("  • yield_change (year-over-year change)")
print("  • years_since_start")

# Check how many NaN values created by lag features
print(f"\nNaN values created by lag features: {df_sorted['yield_lag1'].isna().sum()}")

# Replace df with sorted version
df = df_sorted.copy()
```

## HANDLE NaN VALUES FROM LAG FEATURES

```python
print("\n" + "="*70)
print("🔧 STEP 7: Handle NaN Values")
print("="*70)

print("\nNaN count before handling:")
print(df[['yield_lag1', 'yield_ma2', 'yield_change']].isna().sum())

# Option 1: Fill with median (safe approach)
df['yield_lag1'].fillna(df['yield'].median(), inplace=True)
df['yield_ma2'].fillna(df['yield'].median(), inplace=True)
df['yield_change'].fillna(0, inplace=True)

print("\nNaN count after handling:")
print(df[['yield_lag1', 'yield_ma2', 'yield_change']].isna().sum())

print("\n✓ All NaN values handled")
```

## FEATURE CORRELATION ANALYSIS

```python
print("\n" + "="*70)
print("📊 STEP 8: Feature Correlation with Yield")
print("="*70)

# Select only numeric features
numeric_features = df.select_dtypes(include=[np.number]).columns.tolist()

# Remove target and unnecessary columns
exclude_cols = ['yield', 'production', 'area', 'year', 'fertilizer', 'pesticide']
feature_cols = [col for col in numeric_features if col not in exclude_cols]

# Calculate correlation with yield
correlations = df[feature_cols + ['yield']].corr()['yield'].drop('yield').sort_values(ascending=False)

print("\n🔝 Top 15 Features Correlated with Yield:")
print(correlations.head(15).round(3))

print("\n🔻 Bottom 10 Features (Weakest Correlation):")
print(correlations.tail(10).round(3))

# Visualize top correlations
plt.figure(figsize=(10, 8))
top_features = correlations.head(15)
plt.barh(range(len(top_features)), top_features.values)
plt.yticks(range(len(top_features)), top_features.index)
plt.xlabel('Correlation with Yield')
plt.title('Top 15 Features Correlated with Yield', fontsize=14, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.savefig('stage2_feature_correlations.png', dpi=100, bbox_inches='tight')
print("\n✓ Chart saved: stage2_feature_correlations.png")
plt.show()
```

## FEATURE DISTRIBUTION ANALYSIS

```python
print("\n" + "="*70)
print("🧹 STEP 1: Basic Data Inspection (No Imputation Yet)")
print("="*70)

# Check for missing values (ONLY inspect here)
print("\n🔍 Checking for missing values...")
missing_before = df.isnull().sum()
missing_cols = missing_before[missing_before > 0]

if len(missing_cols) > 0:
    print(f"\n⚠️  Found missing values in {len(missing_cols)} columns:")
    for col, count in missing_cols.items():
        print(f"  • {col}: {count} ({count/len(df)*100:.2f}%)")
else:
    print("✅ No missing values found!")

# Check for infinite values
print("\n🔍 Checking for infinite values...")
numeric_cols = df.select_dtypes(include=[np.number]).columns
inf_counts = {}

for col in numeric_cols:
    inf_count = np.isinf(df[col]).sum()
    if inf_count > 0:
        inf_counts[col] = inf_count

if inf_counts:
    print("⚠️ Found infinite values:")
    for col, count in inf_counts.items():
        print(f"  • {col}: {count}")
else:
    print("✅ No infinite values found!")

print("\n" + "="*70)
print("📋 STEP 2: Prepare Features and Target")
print("="*70)

# Define target variable
target_col = 'yield'

# Columns to drop (non-predictive)
drop_cols = ['yield', 'production', 'area', 'fertilizer', 
             'pesticide', 'rainfall_category', 'pH_category', 'yield_lag1','yield_ma2','yield_change']

drop_cols_exist = [col for col in drop_cols if col in df.columns]

# Prepare X and y
X = df.drop(columns=drop_cols_exist, errors='ignore')
y = df[target_col]

print(f"\n✓ Features (X): {X.shape}")
print(f"✓ Target (y): {y.shape}")

# Identify categorical and numerical features automatically
categorical_features = X.select_dtypes(include=['object']).columns.tolist()
numerical_features = X.select_dtypes(include=[np.number]).columns.tolist()

print(f"\n📊 Feature Breakdown:")
print(f"  Categorical features ({len(categorical_features)}): {categorical_features}")
print(f"  Numerical features ({len(numerical_features)}): {len(numerical_features)}")

print("\n📋 Sample of data:")
print(X.head(3))
```

## CREATE PREPROCESSING PIPELINE

```python
print("\n" + "="*70)
print("🔧 STEP 3: Create Preprocessing Pipeline")
print("="*70)

# Numerical pipeline
numeric_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),   # Prevent leakage
    ('scaler', StandardScaler())
])

# Categorical pipeline
categorical_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# Combine into ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_pipeline, numerical_features),
        ('cat', categorical_pipeline, categorical_features)
    ],
    remainder='drop'
)

print("\n✓ Preprocessor created:")
print("  • Median imputation + StandardScaler for numerical features")
print("  • Mode imputation + OneHotEncoder for categorical features")
```

## TRAIN-TEST SPLIT

```python
print("\n" + "="*70)
print("✂️ STEP 4: Time-Based Split (Train: ≤2017, Test: >2017)")
print("="*70)

# Ensure data is sorted
df = df.sort_values('year').reset_index(drop=True)

# Time-based split
train_df = df[df['year'] <= 2017]
test_df  = df[df['year'] > 2017]

# Separate features and target
X_train = train_df.drop(columns=drop_cols, errors='ignore')
y_train = train_df['yield']

X_test = test_df.drop(columns=drop_cols, errors='ignore')
y_test = test_df['yield']

print("\n✓ Time-based split completed:")
print(f"  Training set: {X_train.shape[0]:,} samples")
print(f"  Test set: {X_test.shape[0]:,} samples")
print(f"  Train years: {train_df['year'].min()} - {train_df['year'].max()}")
print(f"  Test years: {test_df['year'].min()} - {test_df['year'].max()}")
```

## IMPORT MODELS

```python
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, ExtraTreesRegressor
from sklearn.linear_model import Ridge

print("\n✓ Models imported:")
print("  1. Random Forest (your current model)")
print("  2. Gradient Boosting (powerful ensemble)")
print("  3. Extra Trees (faster variant of Random Forest)")
print("  4. Ridge Regression (linear baseline)")
```

## DEFINE MODELS TO TEST

```python
print("\n" + "="*70)
print("🤖 STEP 5: Define Models to Test")
print("="*70)


# Define models inside pipeline (IMPORTANT FIX)
models = {
    'Random Forest': Pipeline([
        ('preprocessing', preprocessor),
        ('model', RandomForestRegressor(
            n_estimators=200,
            max_depth=20,
            min_samples_split=5,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=-1
        ))
    ]),
    
    'Gradient Boosting': Pipeline([
        ('preprocessing', preprocessor),
        ('model', GradientBoostingRegressor(
            n_estimators=200,
            max_depth=7,
            learning_rate=0.1,
            min_samples_split=5,
            min_samples_leaf=2,
            random_state=42
        ))
    ]),
    
    'Extra Trees': Pipeline([
        ('preprocessing', preprocessor),
        ('model', ExtraTreesRegressor(
            n_estimators=200,
            max_depth=20,
            min_samples_split=5,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=-1
        ))
    ]),
    
    'Ridge Regression': Pipeline([
        ('preprocessing', preprocessor),
        ('model', Ridge(
            alpha=1.0,
            random_state=42
        ))
    ])
}

print(f"\n✓ {len(models)} models defined inside proper ML pipelines")
print("\n✓ Each model now includes preprocessing (imputation + encoding + scaling)")
```

## TRAIN AND EVALUATE ALL MODELS

```python
print("\n" + "="*70)
print("🏋️ STEP 6: Training and Evaluating Models")
print("="*70)
print("\nThis may take a few minutes... ⏳\n")

from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Store results
results = {}

# Time-series cross-validation (NO SHUFFLE)
tscv = TimeSeriesSplit(n_splits=5)

for name, model in models.items():
    print(f"{'='*70}")
    print(f"Training: {name}")
    print(f"{'='*70}")
    
    # Cross-validation on training data
    print("  Running TimeSeries cross-validation...")
    
    cv_scores = []
    
    for train_idx, val_idx in tscv.split(X_train):
        X_tr, X_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
        y_tr, y_val = y_train.iloc[train_idx], y_train.iloc[val_idx]
        
        model.fit(X_tr, y_tr)
        val_pred = model.predict(X_val)
        val_mae = mean_absolute_error(y_val, val_pred)
        cv_scores.append(val_mae)
    
    cv_mae = np.mean(cv_scores)
    cv_std = np.std(cv_scores)
    
    # Train on full training set
    print("  Training on full training set...")
    model.fit(X_train, y_train)
    
    # Predict on test set
    print("  Making predictions on test set...")
    y_pred = model.predict(X_test)
    
    # Calculate metrics
    test_mae = mean_absolute_error(y_test, y_pred)
    test_mse = mean_squared_error(y_test, y_pred)
    test_rmse = np.sqrt(test_mse)
    test_r2 = r2_score(y_test, y_pred)
    
    # Store results
    results[name] = {
        'model': model,
        'cv_mae': cv_mae,
        'cv_std': cv_std,
        'test_mae': test_mae,
        'test_rmse': test_rmse,
        'test_r2': test_r2,
        'predictions': y_pred
    }
    
    # Print results
    print(f"\n  ✓ Results for {name}:")
    print(f"     CV MAE: {cv_mae:.2f} (±{cv_std:.2f})")
    print(f"     Test MAE: {test_mae:.2f}")
    print(f"     Test RMSE: {test_rmse:.2f}")
    print(f"     Test R²: {test_r2:.4f}")
    print()

print("\n✅ All models trained successfully!")
```

```python
print("\n" + "="*70)
print("🌾 STEP 2: Crop-wise Performance Analysis (Test Set Only)")
print("="*70)

# Create test results dataframe
test_results = X_test.copy()
test_results['actual_yield'] = y_test.values
test_results['predicted_yield'] = best_predictions
test_results['abs_error'] = np.abs(test_results['actual_yield'] - test_results['predicted_yield'])

# Safe percent error
test_results['percent_error'] = np.where(
    test_results['actual_yield'] != 0,
    (test_results['abs_error'] / test_results['actual_yield']) * 100,
    0
)

# Calculate crop-wise metrics (TEST SET ONLY)
crop_performance = test_results.groupby('crop').agg({
    'abs_error': ['mean', 'median', 'std'],
    'percent_error': 'mean',
    'actual_yield': ['count', 'mean']
}).round(2)

crop_performance.columns = [
    'MAE',
    'Median_Error',
    'Std_Error',
    'Mean_Percent_Error',
    'Count',
    'Avg_Yield'
]

crop_performance = crop_performance.sort_values('MAE', ascending=False)

print("\n📊 Performance by Crop (Top 15 - Highest MAE):")
print(crop_performance.head(15))

print("\n📊 Performance by Crop (Bottom 10 - Lowest MAE):")
print(crop_performance.tail(10))

# Identify problematic crops
threshold = crop_performance['MAE'].quantile(0.75)
problematic_crops = crop_performance[crop_performance['MAE'] > threshold]

print(f"\n⚠️ Crops with High Error (MAE > 75th percentile):")
print(f"  Total: {len(problematic_crops)} crops")

for crop in problematic_crops.head().index:
    print(f"    • {crop}: MAE = {problematic_crops.loc[crop, 'MAE']:.2f}")
```

## VISUALIZE CROP PERFORMANCE

```python
# Visualize crop-wise performance (TEST SET ONLY)
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Plot 1: MAE by crop (top 15)
top_15_crops = crop_performance.head(15)
axes[0, 0].barh(range(len(top_15_crops)), top_15_crops['MAE'].values)
axes[0, 0].set_yticks(range(len(top_15_crops)))
axes[0, 0].set_yticklabels(top_15_crops.index)
axes[0, 0].set_xlabel('Mean Absolute Error')
axes[0, 0].set_title('Top 15 Crops with Highest MAE')
axes[0, 0].grid(axis='x', alpha=0.3)

# Plot 2: Average yield vs MAE
axes[0, 1].scatter(crop_performance['Avg_Yield'],
                   crop_performance['MAE'],
                   alpha=0.6,
                   s=100,
                   c=crop_performance['Count'],
                   cmap='viridis')
axes[0, 1].set_xlabel('Average Yield (kg/ha)')
axes[0, 1].set_ylabel('Mean Absolute Error')
axes[0, 1].set_title('Average Yield vs MAE')
axes[0, 1].grid(alpha=0.3)
cbar = plt.colorbar(axes[0, 1].collections[0], ax=axes[0, 1])
cbar.set_label('Number of Test Records')

# Plot 3: Distribution of errors by crop category
high_yield_crops = crop_performance[crop_performance['Avg_Yield'] > 100].index
medium_yield_crops = crop_performance[
    (crop_performance['Avg_Yield'] > 10) &
    (crop_performance['Avg_Yield'] <= 100)
].index
low_yield_crops = crop_performance[crop_performance['Avg_Yield'] <= 10].index

errors_high = test_results[test_results['crop'].isin(high_yield_crops)]['abs_error']
errors_medium = test_results[test_results['crop'].isin(medium_yield_crops)]['abs_error']
errors_low = test_results[test_results['crop'].isin(low_yield_crops)]['abs_error']

axes[1, 0].boxplot([errors_low, errors_medium, errors_high],
                   labels=['Low (<10)', 'Medium (10–100)', 'High (>100)'])
axes[1, 0].set_ylabel('Absolute Error')
axes[1, 0].set_xlabel('Crop Category by Yield')
axes[1, 0].set_title('Error Distribution by Crop Category')
axes[1, 0].grid(alpha=0.3)

# Plot 4: Percent error distribution (TEST SET ONLY)
axes[1, 1].hist(test_results['percent_error'].clip(0, 50),
                bins=50,
                edgecolor='black',
                alpha=0.7)
axes[1, 1].axvline(test_results['percent_error'].mean(),
                   linestyle='--',
                   linewidth=2,
                   label=f"Mean: {test_results['percent_error'].mean():.1f}%")
axes[1, 1].set_xlabel('Percent Error (%)')
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].set_title('Distribution of Percent Errors (Test Set)')
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('stage4_crop_performance.png', dpi=300, bbox_inches='tight')
print("\n✓ Chart saved: stage4_crop_performance.png")
plt.show()
```
